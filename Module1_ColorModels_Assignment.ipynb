## Q1: RGB to CMYK Conversion
Write a Python program that:
- Takes RGB input values (0-255)
- Converts them to normalized RGB
- Converts RGB to CMYK using standard formulas
- Prints CMYK values rounded to 3 decimal places

**Example Input:** R=255, G=128, B=0
**Expected Output:** `C=0.000 , M=0.498 , Y=1.000 , K=0.000`
---

```
# ✅ Solution for Q1 - RGB to CMYK Conversion
R = int(input('Enter R (0-255): '))
G = int(input('Enter G (0-255): '))
B = int(input('Enter B (0-255): '))

# Normalize RGB
r = R / 255.0
g = G / 255.0
b = B / 255.0

# Convert to CMY
C = 1 - r
M = 1 - g
Y = 1 - b

# Compute K (Black)
K = min(C, M, Y)

if K != 1:
    C = (C - K) / (1 - K)
    M = (M - K) / (1 - K)
    Y = (Y - K) / (1 - K)

print(f'C={C:.3f}, M={M:.3f}, Y={Y:.3f}, K={K:.3f}')
```

## Q2: RGB to HSV Conversion & Visualization
- Load a sample image
- Convert RGB to HSV
- Display both original and HSV images
- Print HSV value of a pixel at (100,50)
---

```
# ✅ Solution for Q2 - RGB to HSV Conversion
import cv2
import matplotlib.pyplot as plt

# Load image
img = cv2.imread('sample.jpg')  # Replace with your image filename
if img is None:
    print('Image not found. Please upload sample.jpg')
else:
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    hsv_img = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)

    # Display images
    plt.figure(figsize=(10,5))
    plt.subplot(1,2,1)
    plt.title('RGB Image')
    plt.imshow(img_rgb)
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.title('HSV Image')
    plt.imshow(hsv_img)
    plt.axis('off')
    plt.show()

    # HSV value of pixel
    print('HSV at (100,50):', hsv_img[50,100])
```

## Q3: Distance Metrics Between Two Pixels
- Calculate Manhattan, Euclidean & Chessboard distances
- Input: two pixel coordinates
---

```
# ✅ Solution for Q3 - Pixel Distance Metrics
import math

x1, y1 = map(int, input('Enter pixel 1 (x1 y1): ').split())
x2, y2 = map(int, input('Enter pixel 2 (x2 y2): ').split())

dx = abs(x1 - x2)
dy = abs(y1 - y2)

manhattan = dx + dy
chessboard = max(dx, dy)
euclidean = math.sqrt(dx**2 + dy**2)

print('Manhattan Distance:', manhattan)
print('Chessboard Distance:', chessboard)
print('Euclidean Distance:', round(euclidean, 3))
```
